# Cleaning

```python
df.dtypes
```

### Member_id

```python
df['member_id'].duplicated().sum() #checking for duplicates
```

### Age

```python
df['age'] = df['age'].astype(float) #changing age type
```

### Gender

```python
df['gender'].unique()
```

```python
gender_dict = {
    "F": "female",
    "M": "male",                #fixing gender labels
    "O": "Other"
               }

df['gender'] = df['gender'].replace(gender_dict)

df['gender'] = df['gender'].str.lower()

df['gender'] = df['gender'].str.strip()

df['gender'].unique()
```

### Industry

```python
df['industry'].unique()
```

```python
industry_dict = {
    "realestate": "real estate",
    "it": "information technology"
}

s = (
    df['industry']
    .str.strip()
    .str.lower()
    .str.replace("/", " & ", regex=False) # "/" -> " & "
    .str.replace(r"\band\b", "&", regex=True) # "and" -> " & "
    .str.replace(r"[-_]", " ", regex=True) # "-" or "_" -> " "
    .str.replace(r"\s*&\s*", " & ", regex=True) # "&" or with more spaces -> " & "
    .str.replace(r"\s+", " ", regex=True) # 1 or more spaces to 1 space
    .str.strip()
)

df['industry'] = s.replace(industry_dict)
df['industry'].unique()
```

### Country

```python
df['country'].unique()
```

```python
country_dict = {
    'portucal' : 'Portugal',
    'portugal ' : 'Portugal',
    'de' : 'Germany',
    'pt' : 'Portugal',
    'deutschland' : 'Germany',
    'uk' : 'United Kingdom',
    'ie' : 'ireland',
    'u.k.' : 'united kingdom',
    'gb' : 'united kingdom',
    'es' : 'spain',
    'españa' : ' spain',
    'espana' : 'spain',
    'fr' : 'france',
    'nl' : 'netherlands',
    'the netherlands' : 'netherlands',
    'holland' : 'netherlands',
}

df['country'] = df['country'].replace(country_dict)

df['country'] = df['country'].str.lower()

df['country'].unique()
```
